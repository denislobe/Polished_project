### Test

```python
test_path = Path("D:\\Datasets\\aluminium_tilt_series\\preprocessed\\mean_area_I")
folder = list_sorted_files(test_path)

hsdata = hs.load(folder[55])

match_disk = hsdata.template_match_disk(disk_r=7, subtract_min=False)
peak_vectors = match_disk.get_diffraction_vectors(method="laplacian_of_gaussian")

mask = peak_vectors.to_mask(disk_r = 2)
center_mask = ~match_disk.get_direct_beam_mask(10)
mask = mask & center_mask
masked_data = hsdata * mask
mask.set_signal_type("electron_diffraction")
scale = hsdata.axes_manager[2].scale
define_calibration(mask, scale)
define_calibration(masked_data, scale)
```

```python
hsdata.plot(cmap='viridis', norm='symlog')
```

```python
mask.plot()
```

```python
masked_data.plot(cmap='viridis', norm='symlog')
```

```python
sim = Sim(angular_resolution=0.5, precession_angle=1.0, min_intensity=1e-5, atom_type="Al", lattice_constant=4.05, space_group=225)
masked_data.change_dtype("float32")
azi = masked_data.get_azimuthal_integral2d(npt=256)
azi = azi ** 0.5
result_data = azi.get_orientation(sim.simulations, n_best=sim.grid.size, frac_keep=1.0).data
```

```python
test_result = TemplateMatching(data=result_data, sim=sim)
```

```python
hsdata.plot(cmap='viridis', norm='symlog')
hsdata.add_marker(test_result.results.to_markers(annotate=True))
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection="ipf", symmetry=test_result.sim.phase.point_group)

template_index = test_result.results.data[0, 0, 0, 0].astype("int16")
optical_axis = Miller(uvw=[0, 0, 1], phase=test_result.sim.phase)
miller = (test_result.sim.simulations.get_simulation(template_index)[0] * optical_axis).round()

cors = test_result.results.data[0, 0, :, 1] / np.max(test_result.results.data[0, 0, :, 1])
ax.scatter(test_result.orientations[0, 0], c=cors, cmap='magma')
ax.scatter(test_result.orientations[0, 0, 0], c="g", s=100, marker="o", label=f"Best match: {miller.uvw[0]}")

```
